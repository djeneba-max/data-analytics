```python
# Transform the below value into a boolean using the bool() function

0
bool_value = bool(0)

print(bool_value)
type(bool_value)
```

```python
# What other number or string value(s) can be typecast as boolean values? Use the print() function to list them
# Numbers
# Python follows a simple rule: 0 is False; all non-zero numbers are True.
# Examples:
print(bool(0))    # False
print(bool(1))    # True
print(bool(5))    # True
print(bool(-3))   # True
print(bool(0.0))  # False
print(bool(2.5))  # True
print(bool(10))   # True


# Strings
# Python follows a simple rule: An empty string ('') is False; any non-empty string is True
# Examples:
print(bool('Hello'))  # True
print(bool('Python')) # True
print(bool('12'))     # True
print(bool(' '))      # True, a space inside quotation marks is a non-empty string.
print(bool(''))       # False
print(bool('0'))      # True
print(bool('False'))  # True
```

## Math with Integers and Floats

### Understanding the difference between `/` and `//` (and how this impacts datatypes)

Write an expression to illustrate the answer to each question below. Make sure to run the cell to check your results.

```python
# What is the result of 2 times 2?

print( 2*2 )
type( 2*2 )
```

```python
# What is the result of 4 divided by 2, using regular division?

print(4 / 2)
type(4 / 2)
```

```python
# What is the result of 4 divided by 2, using floor division?

print(4 // 2)
type(4 // 2)
# The result is now an integer. 
```

```python
# Ok, how about 2 times 2.5?

print(2 *2.5)
type(2 * 2.5)
# Python returns a float because one of the numbers is a float.
```
