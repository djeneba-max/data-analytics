# Exercise 2.A Using built-in functions
## Lab 2. Parsing strings – part codes

```python
def get_supplier_code(part_code):
    return part_code[:part_code.index(':')]
# returns all characters before the :.
# part_code.index(':') finds the position (index) of the ':' character in the part_code string.
```

```python
def get_product_num(part_code):
    start = part_code.index(':') + 1
    end = part_code.index('-')
    return part_code[start:end]
# returns all characters between the : and -.
```

```python
def get_size(part_code):
    return part_code[part_code.index('-') + 1:]
# returns all characters after the -.
```

```python
part_code1 = "ACME:123-L"
part_code2 = "DI:12345-M"
part_code3 = "ACE:1-12"
```

```python
# part_code1 = "ACME:123-L"

supplier = get_supplier_code(part_code1)
product_num = get_product_num(part_code1)
size = get_size(part_code1)

print(f"For {part_code1}, the supplier is {supplier}. Part number {product_num} is available in size {size}.")
```

```python
# part_code2 = "DI:12345-M"

supplier = get_supplier_code(part_code2)
product_num = get_product_num(part_code2)
size = get_size(part_code2)

print(f"For {part_code2}, the supplier is {supplier}. Part number {product_num} is available in size {size}.")
```

```python
# part_code3 = "ACE:1-12"

supplier = get_supplier_code(part_code3)
product_num = get_product_num(part_code3)
size = get_size(part_code3)

print(f"For {part_code3}, the supplier is {supplier}. Part number {product_num} is available in size {size}.")
```

```python
list = ["ACME:123-L","DI:12345-M", "ACE:1-12"]
for part_code in list: 
    supplier = get_supplier_code(part_code)
    product_num = get_product_num(part_code)
    size = get_size(part_code)
    
    print(f"For {part_code}, the supplier is {supplier}. Part number {product_num} is available in size {size}.")

# Indentation is important !
# If print() is inside the loop, it is part of the loop's instructions and runs for every part code.
# If print() is outside the loop, it runs only once, using the last value assigned to part_code.
```
