## Lab 1. Working with names – parsing strings

```python
test_name = 'Maya Lee'
```

```python
def parse_and_display(name):
    for i in range(len(name)):
        if name[i] == ' ':
            first = name[:i]
            last = name[i+1:]
            print(f"Full name: {name}")
            print(f"First name: {first}")
            print(f"Last name: {last}")
            print()
```

```python
parse_and_display(test_name)
```

```python
test_list = ['Grace Flores', 'JB Oinonen', 'Ken Penn']
```

```python
parse_and_display(test_list[0]) # output for Grace Flores
parse_and_display(test_list[1]) # output for JB Oinonen
parse_and_display(test_list[2]) # output for Ken Penn
```

```python
parse_and_display('Djeneba Sacko')
```

```python
def parse_and_display2(name):
    words = name.split()   #split() splits a string into multiple elements using spaces as separators.

    if len(words) == 1:
        first = words[0]
        print(f'Full name: {name}')
        print(f'First name: {first}')
       

    elif len(words) == 2:
        first = words[0]
        last = words [1]
        print(f'Full name: {name}')
        print(f'First name: {first}')
        print(f'Last name: {last}')
        

    elif len(words) == 3:
        first = words[0]
        middle = words[1]
        last = words[2]
        print(f'Full name: {name}')
        print(f'First name: {first}')
        print(f'Middle name: {middle}')
        print(f'Last name: {last}')
        
    print() # Prints a blank line to separate each output.

```

```python
test_list2 = ['Maya', 'Bintou Toure', 'Thierno Baba Ly']
```
