## Working with datetime functions

```python
import datetime
```

# Today

```python
today = datetime.datetime.now()
print(today)
```

```python
print(today.strftime("Today is %A, %B %d, %Y."))
# strftime() converts a date and time into a string (text) in the desired format.
```

# My birthday

```python
birth_date = datetime.date(2000, 11, 15)
print(birth_date)
```

```python
print(birth_date.strftime("My birthday is %m/%d/%Y."))
```

# Next 90 Days

```python
ninety_d = today + datetime.timedelta(days=90)
print(ninety_d)
```

```python
print(ninety_d.strftime("90 days from today is %B %d, %Y."))
```

# Time for Dinner

```python
dinner_time = datetime.time(19, 00)
print(dinner_time)
```

```python
print(dinner_time.strftime("Let's meet for dinner at %I:%M %p."))
```
