## Working with defining custom functions

```python
def favorite_things(name, movie, music):
    return f"My name is {name}.\nMy favorite movie : {movie}.\nMy favorite musician/band : {music}."
    
```

```python
print(favorite_things("Maya", "The Lion King", "N/A"))
```

```python
def why_im_here(purpose):
    return purpose 
```

```python
print(why_im_here("I joined the YUU Data Analyst program to develop my data analytics skills and build a career in the field."))
```

```python
def favorite_place(place, why):
    return f"One of my favorite places is {place} because {why}."
```
