## Working with tax functions

```python
def get_soc_sec_tax(gross_pay):
    return gross_pay * 0.062
```

```python
def get_medicare_tax(gross_pay):
    return gross_pay * 0.0145
```

```python
def get_federal_tax(gross_pay, withholding_code):
    if withholding_code == 0:
        tax_rate = 0.23
    elif withholding_code == 1:
        tax_rate = 0.21
    elif withholding_code == 2:
        tax_rate = 0.195
    elif withholding_code == 3:
        tax_rate = 0.185
    else:
        tax_rate = 0.18 - ((withholding_code - 4) * 0.005)

    return gross_pay * tax_rate
# Starting at code 4, we subtract an additional 0.5% for each increase in the withholding code.
# The difference between the codes tells us how many times to subtract 0.5%.
# To calculate the tax_rate for each code, we use the following formula:
# tax_rate = 0.18 - ((withholding_code - 4) * 0.005)
```

```python
# Person1: gross pay $750 withholding code 0
print(f"Social Security Tax: ${get_soc_sec_tax(750):.2f}")
print()
print(f"Medicare Tax: ${get_medicare_tax(750):.2f}") 
print()
print(f"Federal Tax: ${get_federal_tax(750, 0):.2f}")
```
