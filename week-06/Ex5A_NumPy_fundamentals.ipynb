## Array Basics

```python
fruits = ["apple", "banana", "cherry", "date"]
```

```python
arr_1 = np.array(fruits)
print(arr_1)
```

```python
print(arr_1.dtype)

# The output is <U6.
# U means Unicode (string).
# 6 represents the maximum number of characters in the strings.


#print(type(arr_1)) , <class 'numpy.ndarray'>,  
# type() returns the type of the object: a NumPy array (ndarray).

# .dtype returns the data type of the values contained in the array: Unicode strings.
```

```python
arr_2 = np.array([
    [10, 11, 12],
    [13, 14, 15]
])

print(arr_2)
```

```python
print(arr_2.T)
```

```python
# Using the transpose attribute .T does not modify the original array.
# It returns a transposed view of the array.
# .T switches rows and columns: the array goes from 2 rows x 3 columns to 3 rows x 2 columns.
```

```python
arr_3 = np.array([
    [1, 3, 5],
    [2, 4, 6]
])

print(arr_3)
```

```python
arr_4 = np.array([arr_3[0, 2], arr_3[1, 1]])

print(arr_4)
```

```python
arr_5 = np.ones((2, 2, 2))

print(arr_5)
# np.ones() creates an array filled with 1s.
```
