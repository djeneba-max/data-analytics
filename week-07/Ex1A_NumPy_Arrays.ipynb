 ## Looping Through Arrays

```python
arrA = np.array([
    (11, 12, 13),
    (14, 15, 16 )])
print(arrA)
```

```python
for x in arrA:
    print(x * 2)
```

```python
# The loop iterates through the first dimension of the array. 
# for x in arrA: → retrieves one row at a time (we can think of a 2D array as a list of lists)
# Each row is multiplied by 2 and printed separately. 
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

```python
# for x in arrA: → retrieves one row at a time from the array.
# for y in x: → iterates through each element of the current row individually.
# print(y * 2) → prints each element separately.
```

```python
# First dimension → rows.
# Second dimension → columns (each element within a row).
```

The first loop iterates through the first dimension (rows) of the array. It processes one row at a time, multiplies all its elements by 2, and prints the result as an array.

The second loop uses a nested loop (a loop inside another loop) to iterate through the second dimension (elements within each row). It processes one element at a time, multiplies it by 2, and prints each result individually.

```python
for x in arrA:
    for y in x:
        loop_result = x * 2
```

```python
print(loop_result)
# The inner loop iterates through each element (y) within the current row (x).
# However, loop_result = x * 2 multiplies the entire row, not the individual element.

# At each iteration, loop_result is reassigned with a value. 
# Once the loop finishes, it only contains the result of the last assignment (the last row multiplied by 2).
```

 ## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA), axis=0)

print(joined_array_1)
```

```python
# np.concatenate() → joins two arrays.
# (arrA, arrA) → specifies the two arrays to concatenate.
# axis=0 (default axis) → joins them along the first dimension (rows).
# axis=0 means that the rows of the second array are added after the rows of the first array.
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)
print(joined_array_2) 
```

```python
# axis=1 → joins them along the second dimension (columns).
# Unlike axis=0, which adds rows one below the other (vertical direction), 
# axis=1 adds columns side by side (horizontal direction).
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))

print(stacked_array_1)
```

```python
# np.stack() → stacks two arrays together.
# (joined_array_1, joined_array_1) → specifies the two arrays to stack.
# axis=0 (default) → stacks the arrays along a new first dimension (layers). 

# In the previous examples, concatenate() combined arrays along an existing axis.
# Here, stack() combines arrays along a new axis (adds a dimension, layers).
```

```python
# stacked_array_2 = np.stack((joined_array_1, joined_array_2))

# print(stacked_array_2)

# ValueError: all input arrays must have the same shape.
# np.stack() requires all arrays to have exactly the same shape.
# joined_array_1 has a shape of (4, 3), while joined_array_2 has a shape of (2, 6).
# Since their shapes are different, NumPy cannot stack them together.
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))

print(another_joined_array)
# default axis, axis=0.
```

```python
# another_2 = np.concatenate((joined_array_1, arrA), axis=1)

# print(another_2)

# ValueError: all the input array dimensions except for the concatenation axis must match exactly, 
# but along dimension 0, the array at index 0 has size 4 and the array at index 1 has size 2.

# If we concatenate along the rows (axis=0), the number of rows can be different for each array.
# If we concatenate along the columns (axis=1), the number of columns can be different for each array.
# In other words, the arrays can have different sizes along the concatenation axis, but all other dimensions must have the same size.
# The dimensions that are not being concatenated must have the same size.
```

```python
horizontal_stack = np.hstack((arrA, arrA))
# hstack() → stacks arrays horizontally (adds columns).

vertical_stack = np.vstack((arrA, arrA))
# vstack() → stacks arrays vertically, one below the other (adds rows).

depth_stack = np.dstack((arrA, arrA))
# dstack() → stacks arrays along the depth (adds a third dimension).

print(horizontal_stack)
print()
print(vertical_stack)
print()
print(depth_stack)
```

## Splitting Arrays

```python
splitsky = np.split(joined_array_1, 2)
print(splitsky)
print()
for array in splitsky:
    print(array)
    print()

# The result contains 2 arrays, each with 2 dimensions (2 rows and 3 columns).
# I used a for loop to display the 2 arrays separately for better visualization.
```

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, 'odd')

print(result_1)

# arrA % 2 == 0 → checks whether each element is even.
# arrA → if the condition is True, returns the original value.
# 'odd' → if the condition is False, returns 'odd'.
```

 ## Sorting and Filtering Arrays

```python
arrB = np.array([[7, 6],
                 [13, 5],
                 [7, 3]])

arrB_sorted = np.sort(arrB)

print(arrB_sorted)
# np.sort(arrB) → sorts the elements of each row in ascending order.
```

```python
print(arrB)
print()

z = arrB < 7
#print(z)
#print()

arrC = arrB[z]

print(arrC)
```

```python
# print(arrB) → displays the original array.
# arrB < 7 → checks whether each element is less than 7.
# z → stores the Boolean array (True/False).
# arrB[z] → selects only the elements where the condition is True.
# arrC → stores the filtered values (values less than 7).

# Boolean indexing → allows you to filter an array using a condition.
```

```python
y = arrB >= 7

arrD = arrB[y]

print(arrD)
```
