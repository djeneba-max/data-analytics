```python
import numpy as np
```

## Generating Random Numbers

```python
r_1 = np.random.randint(1, 21, size=2)

print(r_1)
```

```python
r_2 = np.random.randint(1, 21, size=(1, 2))

print(r_2)
```

```python
r_3 = np.random.random(size=(2, 1))

print(r_3)
# np.random.random() generates random floats between 0 (inclusive) and 1 (exclusive).
# size=(2, 1) creates a 2D array with 2 rows and 1 column.
```

```python
r_4 = np.random.randint(20, 101, size=(4, 5))

print(r_4)
```

```python
r_5 = np.random.random(size=(4, 5))

print(r_5)
```

## Generating Random Choices from an Array

```python
students = np.array([
    "Student_1", "Student_2", "Student_3", "Student_4",
    "Student_5", "Student_6", "Student_7", "Student_8",
    "Student_9", "Student_10", "Student_11", "Student_12"
])
#print(students)
```

```python
call_on = np.random.choice(students)

print(call_on)
```

```python
breakout_groups = np.random.choice(students, size=(3, 4), replace=False)

print(breakout_groups)


# Without replace=False, this is not an effective way to assign breakout groups
# because the same student can appear more than once in different groups.
# To make it effective, we must set replace=False so that each student is
# selected only once and each row represents one group.
```

```python
# Assigning probabilities to each student:
# Student_1: 30%
# Student_2: 15%
# Student_3: 0%
# Student_4: 10%
# Student_5: 10%
# Student_6: 5%
# Student_7: 5%
# Student_8: 5%
# Student_9: 5%
# Student_10: 5%
# Student_11: 5%
# Student_12: 5%
# The probabilities add up to 100%.
call_on_weighted = np.random.choice(students, size=5,  p=[
    0.30, 0.15, 0.00, 0.10, 0.10, 0.05,
    0.05, 0.05, 0.05, 0.05, 0.05, 0.05
])

print(call_on_weighted)
```

## Generating a Normal Distribution

```python
bell_curve = np.random.normal(loc=10, scale=5, size=(4, 5))

print(bell_curve)
# np.random.normal() generates random numbers following a normal distribution (bell-shaped curve).
# loc=10 sets the mean (center/peak) to 10.
# scale=5 sets the standard deviation (spread) to 5.
# size=(4, 5) creates a 2D array with 4 rows and 5 columns.
```
