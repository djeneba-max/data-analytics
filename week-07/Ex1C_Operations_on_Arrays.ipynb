```python
import numpy as np
```

 ## Sample Arrays

```python
# r_1: 1D array containing 2 random integers (1–20)
r_1 = np.random.randint(1, 21, size=2)

# r_2: 2D array containing 1 row and 2 columns
r_2 = np.random.randint(1, 21, size=(1, 2))

# r_3: 2D array containing 2 rows and 1 column (random floats)
r_3 = np.random.random(size=(2, 1))

# r_4: 2D array containing 4 rows and 5 columns (integers 20–100)
r_4 = np.random.randint(20, 101, size=(4, 5))

# r_5: 2D array containing 4 rows and 5 columns (random floats)
r_5 = np.random.random(size=(4, 5))
```

 ## Permutations

```python
print(r_4)
```

```python
np.random.shuffle(r_4)
print(r_4)
```

np.random.shuffle() shuffles elements along axis=0 (the first axis).
For a 2D array, it shuffles the rows, not the individual values within each row.

Only the rows were shuffled. The values within each row remained in the same order.

```python
new_shuffle = np.random.permutation(r_4)
print(new_shuffle)
print()
print(r_4)
```

np.random.shuffle(r_4) modifies the original value of the array (r_4).
np.random.permutation(r_4) returns a shuffled copy and leaves the original array unchanged.

I would then use shuffle when I want to randomly rearrange the original array.

I would use permutation when I want to create a shuffled copy while keeping the original array unchanged.

## Mathematical Operations

```python
# Print the starting arrays
print("r_1:", r_1)
print("r_1 shape:", r_1.shape)
print()

print("r_2:", r_2)
print("r_2 shape:", r_2.shape)
print()

# Add r_1 and r_2 using np.add()
result = np.add(r_1, r_2)

# Print the result and its shape
print("Addition result:", result)
print("Result shape:", result.shape)
```

```python
# Print the starting arrays
print("r_1:", r_1)
print("r_1 shape:", r_1.shape)
print()

print("r_3:", r_3)
print("r_3 shape:", r_3.shape)
print()

# Add r_1 and r_3 using np.add()
result = np.add(r_1, r_3)

# Print the result and its shape
print("Addition result:", result)
print("Result shape:", result.shape)
```

Comment: NumPy broadcasts the arrays to produce a result with shape (2, 2), meaning that NumPy automatically adjusts the dimensions of the arrays to perform an operation between them, even if they don't have exactly the same shape.

```python
# Print the starting arrays
print("r_2:", r_2)
print("r_2 shape:", r_2.shape)
print()

print("r_4:", r_4)
print("r_4 shape:", r_4.shape)
print()

# result = np.add(r_2, r_4)

# print("Addition result:", result)
# print("Result shape:", result.shape)

# ValueError: operands could not be broadcast together with shapes (1,2) (4,5) 
# NumPy raises a ValueError because the arrays have incompatible shapes.
# Broadcasting cannot be performed because their dimensions are not compatible.
```

```python
# Add np.array([1]) to r_4
result = np.add(np.array([1]), r_4)

# Print the result and its shape
print("Addition result:", result)
print("Result shape:", result.shape)
```

```python
# Add np.array([[1]]) to r_4
result_1 = np.add(np.array([[1]]), r_4)

print("Result 1:", result_1)
print("Shape:", result_1.shape)
```

```python
# Add np.array([1,1,1,1,1]) to r_4
result_2 = np.add(np.array([1,1,1,1,1]), r_4)

print("Result 2:", result_2)
print("Shape:", result_2.shape)
```

```python
# Add np.array([[1],[1]]) to r_4
# result_3 = np.add(np.array([[1],[1]]), r_4)

# print("Result 3:", result_3)
# print("Shape:", result_3.shape)

# ValueError: operands could not be broadcast together with shapes (2,1) (4,5).
# Broadcasting fails because dimensions 2 and 4 are incompatible.
```

NumPy first checks the number of dimensions of both arrays. If one array has fewer dimensions, NumPy virtually adds dimensions of size 1 to the left until both arrays have the same number of dimensions.

Next, NumPy checks whether the shapes are compatible by comparing the sizes of their dimensions from right to left. The sizes of the compared dimensions must be identical OR one of them must be equal to 1.

If the shapes are compatible, NumPy performs broadcasting. If one dimension has a size of 1, NumPy can virtually repeat its values to match the size of the corresponding dimension in the other array.

Using the add function, each value in one array is added to the corresponding value in the other array.

The shape of the resulting array is determined by the dimensions of both arrays after broadcasting.

## Cumulative Operations

```python
print(r_4)
```

```python
# Print the sum of all values in r_4
print(np.sum(r_4))
```

```python
# Print the sum of elements along axis 0 (down the rows, column by column)
print(np.sum(r_4, axis=0))
```

```python
# Print the sum of elements along axis 1 (across the columns, row by row)
print(np.sum(r_4, axis=1))
```

## Grocery Problem

```python
# Create an array representing the contents of bag1
# 2 vegetables, 3 fruits, 4 boxes of snacks, 0 beverages
bag1 = np.array([2, 3, 4, 0])

print(bag1)
```

```python
# Create an array representing the contents of bag2
# 4 vegetables, 0 fruits, 1 box of snacks, 2 beverages
bag2 = np.array([4, 0, 1, 2])

print(bag2)
```

Comment: Each position represents a food category. The order must remain the same in both arrays to compare them or perform operations between them.

```python
# Calculate the total number of items in each bag
print(f"The first bag contains {np.sum(bag1)} items, and the second bag contains {np.sum(bag2)} items.")
```

```python
# Calculate the total number of items in each category
total = np.add(bag1, bag2)

print(f"There are {total[0]} vegetables, {total[1]} fruits, {total[2]} boxes of snacks, and {total[3]} beverages in the two grocery bags.")
```

## Teamwork

```python
# Create an array containing the number of doors Ali knocked on each day (Monday to Friday)
ali = np.array([5, 4, 6, 4, 2])
```

```python
# Ben knocked on 0 doors on Monday because he was out of town.
ben = np.array([0, 5, 5, 3, 4])
```

```python
# Carla knocked on 0 doors on Wednesday because she was sick.
carla = np.array([3, 6, 0, 3, 4])
```

```python
all_knocks = np.array([ali, ben, carla])
# np.array() combines the three 1D arrays into one 2D array.
# Each row represents one person's values from Monday to Friday.

# Print the resulting array
print(all_knocks)
```

```python
# The total number of doors each person knocked on during the week
person_totals = np.sum(all_knocks, axis=1)

# The total number of doors the group knocked on during the week
group_total = np.sum(all_knocks)

# Print the results
print("Total doors knocked on by each person during the week:", person_totals)
print("Total doors knocked on by the group:", group_total)
```

```python
each_day = np.sum(all_knocks, axis=0)
print(f'Daily total: {each_day}')

cumulative = np.cumsum(each_day)
print(f'Cumulative total: {cumulative}')
```

```python
# 2 doors knocked on = 1 estimated vote.
# 1 door knocked on = 0.5 estimated vote.
estimated_votes = cumulative * 0.5
print(f'Estimated rolling total of votes by day: {estimated_votes}')
```
